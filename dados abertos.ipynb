```python
%sql
CREATE OR REPLACE TABLE silver_fluxo_caixa_eleitoral AS

-- PARTE 1: DESPESAS (Saídas de Caixa)
SELECT 
    'DESPESA' AS TIPO_TRANSACAO,
    SG_UF AS ESTADO,
    NM_UE AS MUNICIPIO,
    DS_CARGO AS CARGO,
    SG_PARTIDO AS PARTIDO,
    NM_CANDIDATO AS CANDIDATO,
    NR_CNPJ_PRESTADOR_CONTA AS CNPJ_CAMPANHA,
    
    NR_CPF_CNPJ_FORNECEDOR AS CPF_CNPJ_CONTRAPARTE,
    NM_FORNECEDOR AS NOME_CONTRAPARTE,
    DS_CNAE_FORNECEDOR AS SETOR_ECONOMICO,
    
    DT_DESPESA AS DATA_TRANSACAO,
    CAST(REPLACE(VR_DESPESA_CONTRATADA, ',', '.') AS DECIMAL(18,2)) AS VALOR_FINANCEIRO
FROM read_files(
    '/Workspace/Repos/GITHUB/dados-abertos-tse/BRONZE/despesas_contratadas_candidatos_2026_*.csv',
    format => 'csv', header => 'true', sep => ';'
)
WHERE VR_DESPESA_CONTRATADA IS NOT NULL

UNION ALL

-- PARTE 2: RECEITAS (Entradas de Caixa)
SELECT 
    'RECEITA' AS TIPO_TRANSACAO,
    SG_UF AS ESTADO,
    NM_UE AS MUNICIPIO,
    DS_CARGO AS CARGO,
    SG_PARTIDO AS PARTIDO,
    NM_CANDIDATO AS CANDIDATO,
    NR_CNPJ_PRESTADOR_CONTA AS CNPJ_CAMPANHA,
    
    NR_CPF_CNPJ_DOADOR AS CPF_CNPJ_CONTRAPARTE,
    NM_DOADOR AS NOME_CONTRAPARTE,
    DS_CNAE_DOADOR AS SETOR_ECONOMICO,
    
    DT_RECEITA AS DATA_TRANSACAO,
    CAST(REPLACE(VR_RECEITA, ',', '.') AS DECIMAL(18,2)) AS VALOR_FINANCEIRO
FROM read_files(
    '/Workspace/Repos/GITHUB/dados-abertos-tse/BRONZE/receitas_candidatos_2026_*.csv',
    format => 'csv', header => 'true', sep => ';'
)
WHERE VR_RECEITA IS NOT NULL;
```
